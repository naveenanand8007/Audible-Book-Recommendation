```python
import re

def extract_genres(text):
    
    # If Ranks and Genre is missing, return missing
    if pd.isna(text):
        return None
    
    # Extract category names after "#number in"
    genres = re.findall(
        r"#\d[\d,]*\s+in\s+([^,#]+(?:;[^#]+)?)(?=,#|\Z)",
        text
    )
    
    # Remove the general Audible category
    genres = [
        genre.strip()
        for genre in genres
        if "Audible Audiobooks & Originals" not in genre
    ]
    
    return genres
```

```python
def extract_genres(text):
    
    if pd.isna(text):
        return None
    
    # Split the text wherever a new ranking starts
    parts = re.split(r",(?=#\d)", text)
    
    genres = []
    
    for part in parts:
        match = re.search(
            r"#\d[\d,]*\s+(?:in\s+)?(.+)",
            part.strip()
        )
        
        if match:
            genre = match.group(1).strip()
            
            # Remove general Audible ranking
            if "Audible Audiobooks & Originals" not in genre:
                genres.append(genre)
    
    return genres
```

```python
def extract_genres(text):

    if pd.isna(text):
        return None

    # Split whenever another ranking starts
    parts = re.split(r",(?=#\d)", text)

    genres = []

    for part in parts:

        match = re.search(
            r"#\d[\d,]*\s+(?:in\s+)?(.+)",
            part.strip()
        )

        if match:

            genre = match.group(1).strip()

            # Remove only the GENERAL Audible category
            if genre.startswith("Audible Audiobooks & Originals"):
                continue

            if genre.startswith("Free in Audible Audiobooks & Originals"):
                continue

            # Remove unnecessary Audible label from useful categories
            genre = re.sub(
                r"\s*\(Audible Audiobooks & Originals\)\s*$",
                "",
                genre
            )

            genres.append(genre.strip())

    return genres
```

```python
# Medium Q4: Which combination of features provides the most accurate recommendations?

# Select columns needed for feature comparison
recommendation_data = merged_data[
    ["Book Name", "Description", "Genres", "Author", "Rating"]
].copy()

print("Original number of books:", len(recommendation_data))

# Check missing values
print("\nMissing values:")
print(recommendation_data.isnull().sum())

# Remove rows that cannot be used for feature comparison
recommendation_data = recommendation_data.dropna(
    subset=["Book Name", "Description", "Genres", "Author", "Rating"]
).copy()

# Remove books with empty genre lists
recommendation_data = recommendation_data[
    recommendation_data["Genres"].apply(
        lambda x: isinstance(x, list) and len(x) > 0
    )
].copy()

# Reset index
recommendation_data = recommendation_data.reset_index(drop=True)

print("\nBooks available for comparison:", len(recommendation_data))

print("\nMissing values after filtering:")
print(recommendation_data[
    ["Book Name", "Description", "Genres", "Author", "Rating"]
].isnull().sum())
```

```python
# Step 2: Create individual feature matrices

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import MultiLabelBinarizer, OneHotEncoder, StandardScaler

# --------------------------------------------------
# 1. DESCRIPTION FEATURES
# --------------------------------------------------

description_vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

description_matrix = description_vectorizer.fit_transform(
    recommendation_data["Description"]
)


# --------------------------------------------------
# 2. GENRE FEATURES
# --------------------------------------------------

genre_encoder = MultiLabelBinarizer()

genre_matrix = genre_encoder.fit_transform(
    recommendation_data["Genres"]
)


# --------------------------------------------------
# 3. AUTHOR FEATURES
# --------------------------------------------------

author_encoder = OneHotEncoder(
    handle_unknown="ignore"
)

author_matrix = author_encoder.fit_transform(
    recommendation_data[["Author"]]
)


# --------------------------------------------------
# 4. RATING FEATURE
# --------------------------------------------------

rating_scaler = StandardScaler()

rating_matrix = rating_scaler.fit_transform(
    recommendation_data[["Rating"]]
)


# Check shapes

print("Number of books:", len(recommendation_data))

print("\nFeature matrix shapes:")
print("Description:", description_matrix.shape)
print("Genre:", genre_matrix.shape)
print("Author:", author_matrix.shape)
print("Rating:", rating_matrix.shape)
```

```python
# Step 3: Create different feature combinations

from scipy.sparse import hstack, csr_matrix

# Convert genre and rating matrices to sparse format
genre_sparse = csr_matrix(genre_matrix)
rating_sparse = csr_matrix(rating_matrix)

# 1. Description only
features_description = description_matrix

# 2. Genre only
features_genre = genre_sparse

# 3. Description + Genre
features_desc_genre = hstack([
    description_matrix,
    genre_sparse
])

# 4. Description + Genre + Author
features_desc_genre_author = hstack([
    description_matrix,
    genre_sparse,
    author_matrix
])

# 5. Description + Genre + Author + Rating
features_all = hstack([
    description_matrix,
    genre_sparse,
    author_matrix,
    rating_sparse
])

print("Feature combination shapes:\n")

print("Description :", features_description.shape)
print("Genre :", features_genre.shape)
print("Description + Genre :", features_desc_genre.shape)
print("Description + Genre + Author :", features_desc_genre_author.shape)
print("Description + Genre + Author + Rating :", features_all.shape)
```

```python
# Step 4: Compare recommendations produced by each feature combination

from sklearn.metrics.pairwise import cosine_similarity


def get_top_recommendations(feature_matrix, book_name, top_n=5):

    # Find the position of the target book
    book_index = recommendation_data[
        recommendation_data["Book Name"] == book_name
    ].index[0]

    # Calculate similarity between this book and all other books
    similarity_scores = cosine_similarity(
        feature_matrix[book_index],
        feature_matrix
    ).flatten()

    # Do not recommend the same book
    similarity_scores[book_index] = -1

    # Get top similar books
    top_indices = similarity_scores.argsort()[-top_n:][::-1]

    results = recommendation_data.loc[
        top_indices,
        [
            "Book Name",
            "Author",
            "Rating",
            "Genres"
        ]
    ].copy()

    results["Similarity"] = similarity_scores[top_indices]

    return results
```

```python
target_book = "The Girl on the Train"


print("\nDESCRIPTION ONLY")
display(
    get_top_recommendations(
        features_description,
        target_book
    )
)


print("\nGENRE ONLY")
display(
    get_top_recommendations(
        features_genre,
        target_book
    )
)


print("\nDESCRIPTION + GENRE")
display(
    get_top_recommendations(
        features_desc_genre,
        target_book
    )
)


print("\nDESCRIPTION + GENRE + AUTHOR")
display(
    get_top_recommendations(
        features_desc_genre_author,
        target_book
    )
)


print("\nDESCRIPTION + GENRE + AUTHOR + RATING")
display(
    get_top_recommendations(
        features_all,
        target_book
    )
)
```

```python
# Step 6: Compare average Top-5 similarity
# across different feature combinations

import numpy as np
from sklearn.metrics.pairwise import cosine_similarity


def average_top5_similarity(feature_matrix):

    # Calculate similarity matrix
    similarity_matrix = cosine_similarity(feature_matrix)

    top5_scores = []

    for i in range(similarity_matrix.shape[0]):

        scores = similarity_matrix[i].copy()

        # Remove similarity with the book itself
        scores[i] = -1

        # Find top 5 recommendations
        top_5 = np.sort(scores)[-5:]

        # Store their average similarity
        top5_scores.append(
            np.mean(top_5)
        )

    return np.mean(top5_scores)


# Calculate average Top-5 similarity
comparison_results = {
    "Description": average_top5_similarity(
        features_description
    ),

    "Genre": average_top5_similarity(
        features_genre
    ),

    "Description + Genre": average_top5_similarity(
        features_desc_genre
    ),

    "Description + Genre + Author": average_top5_similarity(
        features_desc_genre_author
    ),

    "Description + Genre + Author + Rating": average_top5_similarity(
        features_all
    )
}


print("Average Top-5 Similarity:\n")

for feature, score in comparison_results.items():
    print(f"{feature}: {score:.4f}")
```

```python
# Step 7: Visualize average Top-5 similarity
# for different feature combinations

import matplotlib.pyplot as plt

feature_names = list(comparison_results.keys())
similarity_scores = list(comparison_results.values())

plt.figure(figsize=(11, 6))

plt.bar(
    feature_names,
    similarity_scores
)

plt.title(
    "Average Top-5 Similarity by Feature Combination"
)

plt.xlabel("Feature Combination")
plt.ylabel("Average Top-5 Cosine Similarity")

plt.xticks(
    rotation=25,
    ha="right"
)

plt.ylim(0, 0.7)

plt.tight_layout()
plt.show()
```

```python
# Find books that contain at least one science-fiction related genre

def is_science_fiction(genres):
    if not isinstance(genres, list):
        return False
    
    return any(
        "science fiction" in genre.lower()
        or "sci-fi" in genre.lower()
        for genre in genres
    )


# Filter science-fiction books
sci_fi_books = merged_data[
    merged_data["Genres"].apply(is_science_fiction)
].copy()


print("Total science-fiction books:", len(sci_fi_books))

print("\nMissing ratings:")
print(sci_fi_books["Rating"].isna().sum())

print("\nRating statistics:")
print(sci_fi_books["Rating"].describe())
```

```python
# Remove science-fiction books with missing ratings
sci_fi_ranked = sci_fi_books.dropna(
    subset=["Rating"]
).copy()

# Sort by rating from highest to lowest
sci_fi_ranked = sci_fi_ranked.sort_values(
    by="Rating",
    ascending=False
)

# Display top 15 candidates
print("Top-rated science-fiction books:")

display(
    sci_fi_ranked[
        ["Book Name", "Author", "Rating", "Genres"]
    ].head(15)
)
```

```python
# Count how many science-fiction related genres each book has

def sci_fi_relevance(genres):
    if not isinstance(genres, list):
        return 0
    
    return sum(
        1 for genre in genres
        if "science fiction" in genre.lower()
        or "sci-fi" in genre.lower()
    )


sci_fi_ranked["SciFi_Relevance"] = (
    sci_fi_ranked["Genres"].apply(sci_fi_relevance)
)


# Sort first by Rating, then by Sci-Fi relevance
sci_fi_ranked = sci_fi_ranked.sort_values(
    by=["Rating", "SciFi_Relevance"],
    ascending=[False, False]
)


# Inspect top 10
display(
    sci_fi_ranked[
        [
            "Book Name",
            "Author",
            "Rating",
            "SciFi_Relevance",
            "Genres"
        ]
    ].head(10)
)
```

NLP and Clusting

```python
# NLP and Clustering
# Step 1: Prepare book descriptions

description_data = merged_data[
    [
        "Book Name",
        "Author",
        "Description",
        "Rating",
        "Number of Reviews",
        "Genres"
    ]
].copy()

print("Total books:", len(description_data))

description_data = description_data[
    description_data["Description"].str.strip() != ""
].copy()

print("Books with descriptions:", len(description_data))
```

```python
# Step 2: Remove invalid Audible website descriptions

bad_description = description_data[
    "Description"
].str.contains(
    "completing your purchase",
    case=False,
    na=False
)

print("Invalid descriptions:", bad_description.sum())

description_data = description_data[
    ~bad_description
].copy()

description_data = description_data.reset_index(drop=True)

print("Books remaining:", len(description_data))
```

```python
# Step 3: Convert descriptions into numerical features

from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(
    max_features=5000,
    stop_words="english"
)

X = vectorizer.fit_transform(
    description_data["Description"]
)

print("Number of books:", X.shape[0])
print("Number of NLP features:", X.shape[1])
print("Feature matrix shape:", X.shape)
```

```python
# Step 4: View some vocabulary words

words = vectorizer.get_feature_names_out()

print("First 30 words:")
print(words[:30])
```

```python
# Step 5: Apply K-Means clustering

from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42
)

clusters = kmeans.fit_predict(X)

description_data["Cluster"] = clusters

print(description_data["Cluster"].value_counts())
```

```python
# Step 6: Check books in each cluster

for i in range(3):
    print("\nCluster", i)
    print(
        description_data[
            description_data["Cluster"] == i
        ][["Book Name", "Author"]]
        .head(10)
    )
```

```python
# Step 7: Check description length

description_data["Description_Length"] = (
    description_data["Description"]
    .str.split()
    .str.len()
)

print(
    description_data["Description_Length"]
    .describe()
)

print("\nLongest descriptions:")

print(
    description_data[
        ["Book Name", "Description_Length"]
    ]
    .sort_values(
        "Description_Length",
        ascending=False
    )
    .head(10)
)
```

```python
# Step 8: Elbow Method to find K

import matplotlib.pyplot as plt

inertia = []

for k in range(2, 11):
    
    model = KMeans(
        n_clusters=k,
        random_state=42
    )
    
    model.fit(X)
    
    inertia.append(model.inertia_)

plt.plot(
    range(2, 11),
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.show()
```

```python
# Step 9: Apply K-Means with K = 8

kmeans = KMeans(
    n_clusters=8,
    random_state=42
)

description_data["Cluster"] = kmeans.fit_predict(X)

print("Number of books in each cluster:")

print(
    description_data["Cluster"]
    .value_counts()
    .sort_index()
)
```

```python
# Step 10: Check the small clusters

for cluster in range(8):

    cluster_books = description_data[
        description_data["Cluster"] == cluster
    ]

    if len(cluster_books) <= 5:

        print(f"\nCluster {cluster}:")

        print(
            cluster_books[
                ["Book Name", "Description_Length"]
            ].to_string(index=False)
        )
```

```python
# Step 11: View sample books from the two main clusters

for cluster in [1, 5]:

    print(f"\n--- Cluster {cluster} ---")

    print(
        description_data[
            description_data["Cluster"] == cluster
        ][["Book Name", "Author"]]
        .head(15)
        .to_string(index=False)
    )
```

```python
# Step 12: Find important words in each cluster

words = vectorizer.get_feature_names_out()

for i in range(8):

    center = kmeans.cluster_centers_[i]

    top_words = center.argsort()[-10:][::-1]

    print(f"\nCluster {i}:")
    print(words[top_words])
```

```python
# Step 13: Visualize cluster sizes

cluster_counts = (
    description_data["Cluster"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(8, 5))

plt.bar(
    cluster_counts.index,
    cluster_counts.values
)

plt.xlabel("Cluster")
plt.ylabel("Number of Books")
plt.title("Number of Books in Each Cluster")

plt.show()
```

# hybrid recommendation

```python
# Hybrid Recommendation System
# Step 1: Prepare books available in both systems

hybrid_data = recommendation_data.merge(
    description_data[
        ["Book Name", "Author"]
    ],
    on=["Book Name", "Author"],
    how="inner"
)

hybrid_data = hybrid_data.drop_duplicates(
    subset=["Book Name", "Author"]
).reset_index(drop=True)

print("Books available for hybrid recommendation:",
      len(hybrid_data))

print("\nSample books:")
print(
    hybrid_data[
        ["Book Name", "Author", "Genres", "Rating"]
    ].head()
)
```

```python
# Hybrid Recommendation System
# Step 2: Add descriptions

hybrid_data = hybrid_data.merge(
    description_data[
        ["Book Name", "Author", "Description"]
    ],
    on=["Book Name", "Author"],
    how="left"
)

print("Hybrid data shape:", hybrid_data.shape)

print("\nMissing descriptions:")
print(hybrid_data["Description"].isna().sum())

print("\nSample:")
print(
    hybrid_data[
        ["Book Name", "Genres", "Description"]
    ].head()
)
```

```python
# Hybrid Step 2A: Remove duplicate books before merging descriptions

description_unique = description_data[
    ["Book Name", "Author", "Description"]
].drop_duplicates(
    subset=["Book Name", "Author"]
)

hybrid_data = recommendation_data.merge(
    description_unique,
    on=["Book Name", "Author"],
    how="inner"
)

hybrid_data = hybrid_data.drop_duplicates(
    subset=["Book Name", "Author"]
).reset_index(drop=True)

print("Books available for hybrid recommendation:",
      len(hybrid_data))

print("Missing descriptions:",
      hybrid_data["Description"].isna().sum())
```

```python
# Hybrid Step 3: Create genre features

from sklearn.preprocessing import MultiLabelBinarizer

hybrid_mlb = MultiLabelBinarizer()

hybrid_genre_matrix = hybrid_mlb.fit_transform(
    hybrid_data["Genres"]
)

print("Genre feature matrix shape:",
      hybrid_genre_matrix.shape)
```

```python
# Hybrid Step 4: Create description features

from sklearn.feature_extraction.text import CountVectorizer

hybrid_vectorizer = CountVectorizer(
    max_features=5000,
    stop_words="english"
)

hybrid_description_matrix = hybrid_vectorizer.fit_transform(
    hybrid_data["Description"]
)

print(
    "Description feature matrix shape:",
    hybrid_description_matrix.shape
)
```

```python
# Hybrid Step 5: Find selected book

book_name = "The Girl on the Train"

book_index = hybrid_data[
    hybrid_data["Book Name"] == book_name
].index[0]

print("Book:", book_name)
print("Index:", book_index)

print("\nGenres:")
print(hybrid_data.loc[book_index, "Genres"])
```

```python
# Hybrid Step 6: Calculate both similarity scores

from sklearn.metrics.pairwise import cosine_similarity

genre_scores = cosine_similarity(
    hybrid_genre_matrix[book_index].reshape(1, -1),
    hybrid_genre_matrix
).flatten()

description_scores = cosine_similarity(
    hybrid_description_matrix[book_index],
    hybrid_description_matrix
).flatten()

print("Genre similarity scores:", len(genre_scores))
print("Description similarity scores:", len(description_scores))
```

```python
# Hybrid Step 7: Combine the two similarity scores

hybrid_scores = (
    0.5 * genre_scores +
    0.5 * description_scores
)

print("Number of hybrid scores:", len(hybrid_scores))

print("\nFirst 5 hybrid scores:")
print(hybrid_scores[:5])
```

```python
# Hybrid Step 8: Get Top 5 hybrid recommendations

hybrid_results = []

for index, score in enumerate(hybrid_scores):

    if index != book_index:

        hybrid_results.append(
            (index, score)
        )

hybrid_results = sorted(
    hybrid_results,
    key=lambda x: x[1],
    reverse=True
)

top_5_hybrid = hybrid_results[:5]

print("Hybrid Recommendations for:", book_name)

for index, score in top_5_hybrid:

    print(
        hybrid_data.loc[index, "Book Name"],
        "- Hybrid Score:",
        round(score, 2)
    )
```

The hybrid recommendation system combined genre similarity and description similarity with equal weights of 50% each. For The Girl on the Train, the system recommended books that were more relevant to its thriller and crime themes. Compared with the clustering-based recommendations, the hybrid approach produced more appropriate results because it considered both the type of book through genres and the textual content of the description.

# Compare the three systems

```python
# ============================================================
# CONTENT-BASED RECOMMENDATION SYSTEM
# Using Genres + Cosine Similarity
# ============================================================

# Step 1: Prepare data

recommendation_data = merged_data[
    ["Book Name", "Author", "Genres", "Rating"]
].copy()

recommendation_data = recommendation_data.dropna(
    subset=["Genres"]
).copy()

recommendation_data = recommendation_data.reset_index(drop=True)

print("Total books with genres:", len(recommendation_data))


# ------------------------------------------------------------
# Step 2: Convert genre lists into numerical features
# ------------------------------------------------------------

from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()

genre_matrix = mlb.fit_transform(
    recommendation_data["Genres"]
)

print("Genre matrix shape:", genre_matrix.shape)


# ------------------------------------------------------------
# Step 3: Calculate cosine similarity
# ------------------------------------------------------------

from sklearn.metrics.pairwise import cosine_similarity

genre_similarity = cosine_similarity(
    genre_matrix
)

print("Similarity matrix shape:", genre_similarity.shape)


# ------------------------------------------------------------
# Step 4: Select a book
# ------------------------------------------------------------

book_name = "The Girl on the Train"

book_index = recommendation_data[
    recommendation_data["Book Name"] == book_name
].index[0]

print("\nSelected Book:", book_name)

print(
    "Genres:",
    recommendation_data.loc[book_index, "Genres"]
)


# ------------------------------------------------------------
# Step 5: Find Top 5 similar books
# ------------------------------------------------------------

similarity_scores = list(
    enumerate(
        genre_similarity[book_index]
    )
)

similarity_scores = sorted(
    similarity_scores,
    key=lambda x: x[1],
    reverse=True
)

# Remove selected book itself
top_books = [
    (index, score)
    for index, score in similarity_scores
    if index != book_index
][:5]


print("\nTop 5 Content-Based Recommendations:")

for index, score in top_books:

    print(
        recommendation_data.loc[index, "Book Name"],
        "- Similarity:",
        round(score, 2)
    )


# ------------------------------------------------------------
# Step 6: Display recommendation details
# ------------------------------------------------------------

print("\nRecommendation Details:")

for index, score in top_books:

    print(
        "\nBook:",
        recommendation_data.loc[index, "Book Name"]
    )

    print(
        "Genres:",
        recommendation_data.loc[index, "Genres"]
    )

    print(
        "Rating:",
        recommendation_data.loc[index, "Rating"]
    )

    print(
        "Similarity:",
        round(score, 2)
    )


# ------------------------------------------------------------
# Step 7: Average Top-5 similarity
# ------------------------------------------------------------

content_avg = sum(
    score for index, score in top_books
) / len(top_books)

print(
    "\nAverage Top-5 Similarity:",
    round(content_avg, 2)
)


# ============================================================
# EVALUATION
# Precision@5 and Recall@5
# ============================================================

# We define a relevant book as a book that shares
# at least one genre with the selected book.

selected_genres = set(
    recommendation_data.loc[
        book_index,
        "Genres"
    ]
)

relevant_book_names = []

for i in range(len(recommendation_data)):

    if i != book_index:

        other_genres = set(
            recommendation_data.loc[
                i,
                "Genres"
            ]
        )

        common_genres = selected_genres.intersection(
            other_genres
        )

        if len(common_genres) > 0:

            relevant_book_names.append(
                recommendation_data.loc[
                    i,
                    "Book Name"
                ]
            )


print(
    "\nNumber of relevant books:",
    len(relevant_book_names)
)


# ------------------------------------------------------------
# Step 8: Get names of Top-5 recommended books
# ------------------------------------------------------------

content_book_names = []

for index, score in top_books:

    content_book_names.append(
        recommendation_data.loc[
            index,
            "Book Name"
        ]
    )


# ------------------------------------------------------------
# Step 9: Count relevant recommendations
# ------------------------------------------------------------

relevant_recommended = 0

for book in content_book_names:

    if book in relevant_book_names:
        relevant_recommended += 1


# ------------------------------------------------------------
# Step 10: Calculate Precision@5 and Recall@5
# ------------------------------------------------------------

precision_at_5 = (
    relevant_recommended / 5
)

recall_at_5 = (
    relevant_recommended
    / len(relevant_book_names)
)


# ------------------------------------------------------------
# Step 11: Print final evaluation
# ------------------------------------------------------------

print("\nContent-Based Evaluation")

print(
    "Relevant books recommended:",
    relevant_recommended
)

print(
    "Precision@5:",
    round(precision_at_5, 3)
)

print(
    "Recall@5:",
    round(recall_at_5, 3)
)

print(
    "Average Top-5 Similarity:",
    round(content_avg, 2)
)

print("\nRecommended Books:")

for book in content_book_names:
    print("-", book)
```

```python
# ============================================================
# CLUSTERING-BASED EVALUATION
# Precision@5 and Recall@5
# ============================================================

# Get Top-5 clustering-based recommended book names

cluster_book_names = []

for index, score in top_5_cluster:
    cluster_book_names.append(
        description_data.loc[index, "Book Name"]
    )


# Count how many recommended books are relevant

cluster_relevant_recommended = 0

for book in cluster_book_names:
    if book in relevant_book_names:
        cluster_relevant_recommended += 1


# Calculate Precision@5 and Recall@5

cluster_precision_at_5 = (
    cluster_relevant_recommended / 5
)

cluster_recall_at_5 = (
    cluster_relevant_recommended
    / len(relevant_book_names)
)


# Average similarity

cluster_avg = sum(
    score for index, score in top_5_cluster
) / len(top_5_cluster)


# Final results

print("Clustering-Based Evaluation")

print(
    "Relevant books recommended:",
    cluster_relevant_recommended
)

print(
    "Precision@5:",
    round(cluster_precision_at_5, 3)
)

print(
    "Recall@5:",
    round(cluster_recall_at_5, 3)
)

print(
    "Average Top-5 Similarity:",
    round(cluster_avg, 2)
)

print("\nRecommended Books:")

for book in cluster_book_names:
    print("-", book)
```

```python
# ============================================================
# HYBRID RECOMMENDATION EVALUATION
# Precision@5 and Recall@5
# ============================================================

# Get Top-5 hybrid recommended book names

hybrid_book_names = []

for index, score in top_5_hybrid:
    hybrid_book_names.append(
        hybrid_data.loc[index, "Book Name"]
    )


# Count relevant recommendations

hybrid_relevant_recommended = 0

for book in hybrid_book_names:
    if book in relevant_book_names:
        hybrid_relevant_recommended += 1


# Calculate Precision@5 and Recall@5

hybrid_precision_at_5 = (
    hybrid_relevant_recommended / 5
)

hybrid_recall_at_5 = (
    hybrid_relevant_recommended
    / len(relevant_book_names)
)


# Calculate average Top-5 hybrid score

hybrid_avg = sum(
    score for index, score in top_5_hybrid
) / len(top_5_hybrid)


# Print results

print("Hybrid Recommendation Evaluation")

print(
    "Relevant books recommended:",
    hybrid_relevant_recommended
)

print(
    "Precision@5:",
    round(hybrid_precision_at_5, 3)
)

print(
    "Recall@5:",
    round(hybrid_recall_at_5, 3)
)

print(
    "Average Top-5 Similarity:",
    round(hybrid_avg, 2)
)

print("\nRecommended Books:")

for book in hybrid_book_names:
    print("-", book)
```

```python
# ============================================================
# FINAL FAIR EVALUATION
# Same relevant-book set for all 3 recommendation models
# ============================================================

book_name = "The Girl on the Train"

# Find selected book inside hybrid_data
evaluation_book_index = hybrid_data[
    hybrid_data["Book Name"] == book_name
].index[0]

# Selected book genres
selected_genres = set(
    hybrid_data.loc[
        evaluation_book_index,
        "Genres"
    ]
)

# ------------------------------------------------------------
# Create ONE fixed relevant-book set
# ------------------------------------------------------------

relevant_book_names_fixed = []

for i in range(len(hybrid_data)):

    if i != evaluation_book_index:

        other_genres = set(
            hybrid_data.loc[i, "Genres"]
        )

        if len(
            selected_genres.intersection(other_genres)
        ) > 0:

            relevant_book_names_fixed.append(
                hybrid_data.loc[i, "Book Name"]
            )

print(
    "Total relevant books used for evaluation:",
    len(relevant_book_names_fixed)
)


# ============================================================
# 1. CONTENT-BASED
# ============================================================

content_book_names = [
    recommendation_data.loc[index, "Book Name"]
    for index, score in top_books
]

content_relevant = sum(
    book in relevant_book_names_fixed
    for book in content_book_names
)

content_precision = content_relevant / 5
content_recall = (
    content_relevant /
    len(relevant_book_names_fixed)
)


# ============================================================
# 2. CLUSTERING-BASED
# ============================================================

cluster_book_names = [
    description_data.loc[index, "Book Name"]
    for index, score in top_5_cluster
]

cluster_relevant = sum(
    book in relevant_book_names_fixed
    for book in cluster_book_names
)

cluster_precision = cluster_relevant / 5
cluster_recall = (
    cluster_relevant /
    len(relevant_book_names_fixed)
)


# ============================================================
# 3. HYBRID
# ============================================================

hybrid_book_names = [
    hybrid_data.loc[index, "Book Name"]
    for index, score in top_5_hybrid
]

hybrid_relevant = sum(
    book in relevant_book_names_fixed
    for book in hybrid_book_names
)

hybrid_precision = hybrid_relevant / 5
hybrid_recall = (
    hybrid_relevant /
    len(relevant_book_names_fixed)
)


# ============================================================
# FINAL RESULTS
# ============================================================

print("\nContent-Based")
print("Relevant Recommended:", content_relevant)
print("Precision@5:", round(content_precision, 3))
print("Recall@5:", round(content_recall, 3))

print("\nClustering-Based")
print("Relevant Recommended:", cluster_relevant)
print("Precision@5:", round(cluster_precision, 3))
print("Recall@5:", round(cluster_recall, 3))

print("\nHybrid")
print("Relevant Recommended:", hybrid_relevant)
print("Precision@5:", round(hybrid_precision, 3))
print("Recall@5:", round(hybrid_recall, 3))
```
